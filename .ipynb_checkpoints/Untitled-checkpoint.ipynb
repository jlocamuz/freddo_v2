```python
import os, math, requests, pandas as pd
from concurrent.futures import ThreadPoolExecutor, as_completed

BASE = "https://api-prod.humand.co/public/api/v1"
AUTH = "Basic Njc3NDUzMjpkNWR2Z1pzNXQ3ZEZ2XzE2Z2pfbV9XNklpVFNPU0NmMQ=="

START_DATE = "2025-11-17"
END_DATE   = "2025-12-16"

LIMIT_USERS = 50
LIMIT_DAYS  = 500
BATCH_SIZE  = 25
MAX_WORKERS = 8

# ================= SESSION =================
s = requests.Session()
s.headers.update({"Authorization": AUTH})

def get(url, params):
    r = s.get(url, params=params, timeout=60)
    r.raise_for_status()
    return r.json()

def flatten(d, p=""):
    o = {}
    for k, v in d.items():
        k2 = f"{p}.{k}" if p else k
        if isinstance(v, dict):
            o.update(flatten(v, k2))
        elif isinstance(v, list):
            o[k2] = str(v)
        else:
            o[k2] = v
    return o

# ================= USERS =================
first = get(f"{BASE}/users", {"page": 1, "limit": LIMIT_USERS})
pages = math.ceil(first["count"] / LIMIT_USERS)

users = first["users"]
for p in range(2, pages + 1):
    users += get(f"{BASE}/users", {"page": p, "limit": LIMIT_USERS})["users"]

employee_ids = [
    u["employeeInternalId"]
    for u in users
    if u["status"] == "ACTIVE"
]

print(f"Usuarios ACTIVE: {len(employee_ids)}")

# ================= DAY SUMMARIES (COMMA-SEPARATED) =================
def fetch_batch(emp_ids):
    rows = []
    page = 1
    emp_param = ",".join(emp_ids)   # 👈 CLAVE

    while True:
        data = get(
            f"{BASE}/time-tracking/day-summaries",
            {
                "employeeIds": emp_param,
                "startDate": START_DATE,
                "endDate": END_DATE,
                "limit": LIMIT_DAYS,
                "page": page,
            }
        )

        items = data.get("items", [])
        if not items:
            break

        for it in items:
            rows.append(flatten(it))

        if len(items) < LIMIT_DAYS:
            break

        page += 1

    return rows

# ================= EXEC =================
all_rows = []

with ThreadPoolExecutor(max_workers=MAX_WORKERS) as ex:
    futures = [
        ex.submit(fetch_batch, employee_ids[i:i+BATCH_SIZE])
        for i in range(0, len(employee_ids), BATCH_SIZE)
    ]

    for f in as_completed(futures):
        all_rows.extend(f.result())

print(f"TOTAL FILAS: {len(all_rows)}")

df = pd.DataFrame(all_rows)
df = df.sort_values(
    by=["employeeId", "referenceDate"],  # si la columna se llama "date"
    ascending=[True, True]
)
#df.to_excel("day_summaries_2025-11-17_a_2025-12-16.xlsx", index=False)
```

```python
import os, math, requests, pandas as pd
from concurrent.futures import ThreadPoolExecutor, as_completed
from datetime import datetime
from zoneinfo import ZoneInfo

BASE = "https://api-prod.humand.co/public/api/v1"
AUTH = "Basic Njc3NDUzMjpkNWR2Z1pzNXQ3ZEZ2XzE2Z2pfbV9XNklpVFNPU0NmMQ=="

START_DATE = "2025-11-17"
END_DATE   = "2025-12-16"

LIMIT_USERS = 50
LIMIT_DAYS  = 500
BATCH_SIZE  = 25
MAX_WORKERS = 8

TZ_AR = ZoneInfo("America/Argentina/Buenos_Aires")
DAYS_ES = ["Lunes","Martes","Miércoles","Jueves","Viernes","Sábado","Domingo"]

# ================= SESSION =================
s = requests.Session()
s.headers.update({"Authorization": AUTH})

def get(url, params):
    r = s.get(url, params=params, timeout=60)
    r.raise_for_status()
    return r.json()

# ======= FORMATEOS QUE NECESITÁS =======
def weekday_es(date_str):
    # date_str: "2025-12-02"
    try:
        d = datetime.strptime(date_str[:10], "%Y-%m-%d")
        return DAYS_ES[d.weekday()]
    except Exception:
        return ""

def horario_obligatorio(timeslots):
    # timeslots: [{'startTime':'12:00','endTime':'18:00'}] o None/vacío
    if not timeslots or not isinstance(timeslots, list):
        return ""
    ts0 = timeslots[0] or {}
    st = ts0.get("startTime")
    en = ts0.get("endTime")
    if st and en:
        return f"{st} - {en}"
    return ""

def join_incidences(incidences):
    # incidences: ["ABSENT","LATE"] -> "ABSENT, LATE"
    if not incidences or not isinstance(incidences, list):
        return ""
    return ", ".join([str(x) for x in incidences if x is not None])

def iso_to_ar_hhmm(iso_str):
    # "2025-12-03T05:40:16.126Z" -> "02:40" (AR)
    if not iso_str:
        return ""
    try:
        s = iso_str.replace("Z", "+00:00")
        dt = datetime.fromisoformat(s)
        dt_ar = dt.astimezone(TZ_AR)
        return dt_ar.strftime("%H:%M")
    except Exception:
        return ""

def fichadas_start_end(entries):
    # entries: lista dicts con type START/END y time/date
    if not entries or not isinstance(entries, list):
        return ""
    start_t = ""
    end_t = ""

    # busco primer START y primer END (como venías haciendo)
    for e in entries:
        if not isinstance(e, dict):
            continue
        t = e.get("type")
        iso = e.get("time") or e.get("date")
        if t == "START" and not start_t:
            start_t = iso_to_ar_hhmm(iso)
        elif t == "END" and not end_t:
            end_t = iso_to_ar_hhmm(iso)

    if start_t and end_t:
        return f"{start_t} - {end_t}"
    if start_t:
        return f"{start_t} -"
    if end_t:
        return f"- {end_t}"
    return ""

def regular_extra_from_categorized(categorized):
    # categorized: [{'hours':4,'category':{'name':'REGULAR'}}, ...]
    reg = 0.0
    ext = 0.0
    if not categorized or not isinstance(categorized, list):
        return reg, ext

    for ch in categorized:
        if not isinstance(ch, dict):
            continue
        h = float(ch.get("hours") or 0)
        name = ((ch.get("category") or {}).get("name") or "").upper()
        if name == "REGULAR":
            reg += h
        elif name == "EXTRA":
            ext += h

    return reg, ext

# ================= USERS =================
first = get(f"{BASE}/users", {"page": 1, "limit": LIMIT_USERS})
pages = math.ceil(first["count"] / LIMIT_USERS)

users = first["users"]
for p in range(2, pages + 1):
    users += get(f"{BASE}/users", {"page": p, "limit": LIMIT_USERS})["users"]

# mapa: employeeInternalId -> (firstName, lastName)
user_map = {}
employee_ids = []
for u in users:
    if u.get("status") != "ACTIVE":
        continue
    emp = u.get("employeeInternalId")
    if not emp:
        continue
    employee_ids.append(emp)
    user_map[emp] = {
        "employeeInternalId": emp,
        "firstName": u.get("firstName", "") or "",
        "lastName":  u.get("lastName", "") or "",
    }

print(f"Usuarios ACTIVE: {len(employee_ids)}")

# ================= DAY SUMMARIES =================
def fetch_batch(emp_ids):
    rows = []
    page = 1
    emp_param = ",".join(emp_ids)

    while True:
        data = get(
            f"{BASE}/time-tracking/day-summaries",
            {
                "employeeIds": emp_param,
                "startDate": START_DATE,
                "endDate": END_DATE,
                "limit": LIMIT_DAYS,
                "page": page,
            }
        )

        items = data.get("items", [])
        if not items:
            break

        for it in items:
            emp = it.get("employeeId")  # ojo: en day-summaries se llama employeeId
            ref = (it.get("referenceDate") or it.get("date") or "")[:10]

            # users data
            u = user_map.get(emp, {"firstName": "", "lastName": ""})

            # horario obligatorio
            hor = horario_obligatorio(it.get("timeSlots"))

            # observaciones
            obs = join_incidences(it.get("incidences"))

            # fichadas
            fich = fichadas_start_end(it.get("entries"))

            # horas regulares / extra
            reg, ext = regular_extra_from_categorized(it.get("categorizedHours"))

            rows.append({
                "employeeInternalId": emp,
                "Nombre": u.get("firstName", ""),
                "Apellido": u.get("lastName", ""),
                "weekday": weekday_es(ref),
                "referenceDate": ref,
                "Horario obligatorio": hor,
                "Observaciones": obs,
                "Fichadas": fich,
                "Horas Regulares": reg,
                "Horas Extra": ext,
            })

        if len(items) < LIMIT_DAYS:
            break
        page += 1

    return rows

# ================= EXEC =================
all_rows = []
with ThreadPoolExecutor(max_workers=MAX_WORKERS) as ex:
    futures = [
        ex.submit(fetch_batch, employee_ids[i:i+BATCH_SIZE])
        for i in range(0, len(employee_ids), BATCH_SIZE)
    ]
    for f in as_completed(futures):
        all_rows.extend(f.result())

print(f"TOTAL FILAS: {len(all_rows)}")

df = pd.DataFrame(all_rows).sort_values(
    by=["employeeInternalId", "referenceDate"],
    ascending=[True, True]
)

out = f"reporte_limpio_{START_DATE}_a_{END_DATE}.xlsx"
```
