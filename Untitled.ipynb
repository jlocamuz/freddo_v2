```python
import os, math, requests, pandas as pd
from concurrent.futures import ThreadPoolExecutor, as_completed
from datetime import datetime
from zoneinfo import ZoneInfo

BASE = "https://api-prod.humand.co/public/api/v1"
AUTH = "Basic Njc3NDUzMjpkNWR2Z1pzNXQ3ZEZ2XzE2Z2pfbV9XNklpVFNPU0NmMQ=="

START_DATE = "2025-11-17"
END_DATE   = "2025-12-16"

LIMIT_USERS = 50
LIMIT_DAYS  = 500
BATCH_SIZE  = 25
MAX_WORKERS = 8

TZ_AR = ZoneInfo("America/Argentina/Buenos_Aires")
DAYS_ES = ["Lunes","Martes","Miércoles","Jueves","Viernes","Sábado","Domingo"]

# ================= SESSION =================
s = requests.Session()
s.headers.update({"Authorization": AUTH})

def get(url, params):
    r = s.get(url, params=params, timeout=60)
    r.raise_for_status()
    return r.json()

# ======= FORMATEOS QUE NECESITÁS =======
def weekday_es(date_str):
    # date_str: "2025-12-02"
    try:
        d = datetime.strptime(date_str[:10], "%Y-%m-%d")
        return DAYS_ES[d.weekday()]
    except Exception:
        return ""

def horario_obligatorio(timeslots):
    # timeslots: [{'startTime':'12:00','endTime':'18:00'}] o None/vacío
    if not timeslots or not isinstance(timeslots, list):
        return ""
    ts0 = timeslots[0] or {}
    st = ts0.get("startTime")
    en = ts0.get("endTime")
    if st and en:
        return f"{st} - {en}"
    return ""

def join_incidences(incidences):
    # incidences: ["ABSENT","LATE"] -> "ABSENT, LATE"
    if not incidences or not isinstance(incidences, list):
        return ""
    return ", ".join([str(x) for x in incidences if x is not None])

def iso_to_ar_hhmm(iso_str):
    # "2025-12-03T05:40:16.126Z" -> "02:40" (AR)
    if not iso_str:
        return ""
    try:
        s = iso_str.replace("Z", "+00:00")
        dt = datetime.fromisoformat(s)
        dt_ar = dt.astimezone(TZ_AR)
        return dt_ar.strftime("%H:%M")
    except Exception:
        return ""

def fichadas_start_end(entries):
    # entries: lista dicts con type START/END y time/date
    if not entries or not isinstance(entries, list):
        return ""
    start_t = ""
    end_t = ""

    # busco primer START y primer END (como venías haciendo)
    for e in entries:
        if not isinstance(e, dict):
            continue
        t = e.get("type")
        iso = e.get("time") or e.get("date")
        if t == "START" and not start_t:
            start_t = iso_to_ar_hhmm(iso)
        elif t == "END" and not end_t:
            end_t = iso_to_ar_hhmm(iso)

    if start_t and end_t:
        return f"{start_t} - {end_t}"
    if start_t:
        return f"{start_t} -"
    if end_t:
        return f"- {end_t}"
    return ""

def regular_extra_from_categorized(categorized):
    # categorized: [{'hours':4,'category':{'name':'REGULAR'}}, ...]
    reg = 0.0
    ext = 0.0
    if not categorized or not isinstance(categorized, list):
        return reg, ext

    for ch in categorized:
        if not isinstance(ch, dict):
            continue
        h = float(ch.get("hours") or 0)
        name = ((ch.get("category") or {}).get("name") or "").upper()
        if name == "REGULAR":
            reg += h
        elif name == "EXTRA":
            ext += h

    return reg, ext

# ================= USERS =================
first = get(f"{BASE}/users", {"page": 1, "limit": LIMIT_USERS})
pages = math.ceil(first["count"] / LIMIT_USERS)

users = first["users"]
for p in range(2, pages + 1):
    users += get(f"{BASE}/users", {"page": p, "limit": LIMIT_USERS})["users"]

# mapa: employeeInternalId -> (firstName, lastName)
user_map = {}
employee_ids = []
for u in users:
    if u.get("status") != "ACTIVE":
        continue
    emp = u.get("employeeInternalId")
    if not emp:
        continue
    employee_ids.append(emp)
    user_map[emp] = {
        "employeeInternalId": emp,
        "firstName": u.get("firstName", "") or "",
        "lastName":  u.get("lastName", "") or "",
    }

print(f"Usuarios ACTIVE: {len(employee_ids)}")

# ================= DAY SUMMARIES =================
def fetch_batch(emp_ids):
    rows = []
    page = 1
    emp_param = ",".join(emp_ids)

    while True:
        data = get(
            f"{BASE}/time-tracking/day-summaries",
            {
                "employeeIds": emp_param,
                "startDate": START_DATE,
                "endDate": END_DATE,
                "limit": LIMIT_DAYS,
                "page": page,
            }
        )

        items = data.get("items", [])
        if not items:
            break

        for it in items:
            emp = it.get("employeeId")  # ojo: en day-summaries se llama employeeId
            ref = (it.get("referenceDate") or it.get("date") or "")[:10]

            # users data
            u = user_map.get(emp, {"firstName": "", "lastName": ""})

            # horario obligatorio
            hor = horario_obligatorio(it.get("timeSlots"))

            # observaciones
            obs = join_incidences(it.get("incidences"))

            # fichadas
            fich = fichadas_start_end(it.get("entries"))

            # horas regulares / extra
            reg, ext = regular_extra_from_categorized(it.get("categorizedHours"))

            rows.append({
                "employeeInternalId": emp,
                "Nombre": u.get("firstName", ""),
                "Apellido": u.get("lastName", ""),
                "weekday": weekday_es(ref),
                "referenceDate": ref,
                "Horario obligatorio": hor,
                "Observaciones": obs,
                "Fichadas": fich,
                "Horas Regulares": reg,
                "Horas Extra": ext,
            })

        if len(items) < LIMIT_DAYS:
            break
        page += 1

    return rows

# ================= EXEC =================
all_rows = []
with ThreadPoolExecutor(max_workers=MAX_WORKERS) as ex:
    futures = [
        ex.submit(fetch_batch, employee_ids[i:i+BATCH_SIZE])
        for i in range(0, len(employee_ids), BATCH_SIZE)
    ]
    for f in as_completed(futures):
        all_rows.extend(f.result())

print(f"TOTAL FILAS: {len(all_rows)}")

df = pd.DataFrame(all_rows).sort_values(
    by=["employeeInternalId", "referenceDate"],
    ascending=[True, True]
)

out = f"reporte_limpio_{START_DATE}_a_{END_DATE}.xlsx"
```

```python
import math, requests, pandas as pd
from concurrent.futures import ThreadPoolExecutor, as_completed
from datetime import datetime, timedelta
from zoneinfo import ZoneInfo

# ================= CONFIG =================
BASE = "https://api-prod.humand.co/public/api/v1"
AUTH = "Basic Njc3NDUzMjpkNWR2Z1pzNXQ3ZEZ2XzE2Z2pfbV9XNklpVFNPU0NmMQ=="

START_DATE = "2025-10-01"
END_DATE   = "2025-10-31"

LIMIT_USERS = 50
LIMIT_DAYS  = 500
BATCH_SIZE  = 25
MAX_WORKERS = 8

TZ_AR = ZoneInfo("America/Argentina/Buenos_Aires")
DAYS_ES = ["Lunes","Martes","Miércoles","Jueves","Viernes","Sábado","Domingo"]

# ✅ FLAG: True -> muestra tardanza/anticipado/retiro en HH:MM (00:21)
#          False -> muestra en horas decimales (0.35)
MOSTRAR_METRICAS_EN_HHMM = True

# ================= SESSION =================
s = requests.Session()
s.headers.update({"Authorization": AUTH})

def get(url, params):
    r = s.get(url, params=params, timeout=60)
    r.raise_for_status()
    return r.json()

def weekday_es(date_str_yyyy_mm_dd: str) -> str:
    try:
        d = datetime.strptime(date_str_yyyy_mm_dd, "%Y-%m-%d")
        return DAYS_ES[d.weekday()]
    except Exception:
        return ""

def iso_to_dt_ar(iso_str):
    if not iso_str:
        return pd.NaT
    try:
        return datetime.fromisoformat(iso_str.replace("Z", "+00:00")).astimezone(TZ_AR)
    except Exception:
        return pd.NaT

def fmt_range(start_dt, end_dt):
    if pd.isna(start_dt) or pd.isna(end_dt):
        return ""
    try:
        return f"{start_dt.strftime('%H:%M')} - {end_dt.strftime('%H:%M')}"
    except Exception:
        return ""

def hours_regular_extra(categorized_hours):
    reg = 0.0
    ext = 0.0
    for ch in categorized_hours or []:
        try:
            h = float(ch.get("hours") or 0)
            name = ((ch.get("category") or {}).get("name") or "").upper()
            if name == "REGULAR":
                reg += h
            elif name == "EXTRA":
                ext += h
        except Exception:
            continue
    return reg, ext

def build_observaciones(it: dict) -> str:
    parts = []

    incid = it.get("incidences") or []
    if isinstance(incid, list) and len(incid) > 0:
        parts.append("INCIDENCES: " + ", ".join([str(x) for x in incid if x is not None]))

    tor = it.get("timeOffRequests") or []
    if isinstance(tor, list) and len(tor) > 0:
        parts.append(f"TIME_OFF: {len(tor)}")

    hol = it.get("holidays") or []
    if isinstance(hol, list) and len(hol) > 0:
        names = []
        for h in hol:
            if isinstance(h, dict):
                names.append(h.get("name") or h.get("title") or "")
            elif isinstance(h, str):
                names.append(h)
        names = [n for n in names if n]
        parts.append("HOLIDAY: " + (", ".join(names) if names else str(len(hol))))

    is_workday = it.get("isWorkday")
    if is_workday is False:
        parts.append("NO_WORKDAY")

    return " | ".join(parts)

def horas_a_hhmm(horas):
    """Convierte horas decimales a HH:MM. Ej 0.35 -> 00:21"""
    if pd.isna(horas):
        return ""
    try:
        total_min = int(round(float(horas) * 60))
        h = total_min // 60
        m = total_min % 60
        return f"{h:02d}:{m:02d}"
    except Exception:
        return ""

# ================= USERS =================
first = get(f"{BASE}/users", {"page": 1, "limit": LIMIT_USERS})
pages = math.ceil(first["count"] / LIMIT_USERS)

users = first["users"]
for p in range(2, pages + 1):
    users += get(f"{BASE}/users", {"page": p, "limit": LIMIT_USERS})["users"]

user_map = {}
employee_ids = []

for u in users:
    if u.get("status") != "ACTIVE":
        continue
    emp = u.get("employeeInternalId")
    if not emp:
        continue
    employee_ids.append(emp)
    user_map[emp] = {
        "Nombre": (u.get("firstName") or "").strip(),
        "Apellido": (u.get("lastName") or "").strip()
    }

print(f"Usuarios ACTIVE: {len(employee_ids)}")

# ================= DAY SUMMARIES =================
def fetch_batch(emp_ids):
    rows = []
    page = 1

    while True:
        data = get(
            f"{BASE}/time-tracking/day-summaries",
            {
                "employeeIds": ",".join(emp_ids),
                "startDate": START_DATE,
                "endDate": END_DATE,
                "limit": LIMIT_DAYS,
                "page": page
            }
        )

        items = data.get("items", [])
        if not items:
            break

        for it in items:
            emp = it.get("employeeId")
            ref = (it.get("referenceDate") or it.get("date") or "")[:10]
            u = user_map.get(emp, {"Nombre": "", "Apellido": ""})

            # ===== HORARIO OBLIGATORIO =====
            sched_start = pd.NaT
            sched_end   = pd.NaT
            ts0 = (it.get("timeSlots") or [{}])[0]

            if isinstance(ts0, dict) and ts0.get("startTime"):
                try:
                    h, m = map(int, ts0["startTime"].split(":"))
                    d = datetime.strptime(ref, "%Y-%m-%d")
                    sched_start = datetime(d.year, d.month, d.day, h, m, tzinfo=TZ_AR)
                except Exception:
                    sched_start = pd.NaT

            if isinstance(ts0, dict) and ts0.get("endTime"):
                try:
                    h, m = map(int, ts0["endTime"].split(":"))
                    d = datetime.strptime(ref, "%Y-%m-%d")
                    sched_end = datetime(d.year, d.month, d.day, h, m, tzinfo=TZ_AR)
                    if ts0["endTime"] == "00:00":
                        sched_end += timedelta(days=1)
                    if not pd.isna(sched_start) and sched_end < sched_start:
                        sched_end += timedelta(days=1)
                except Exception:
                    sched_end = pd.NaT

            # ===== HORARIO REAL =====
            real_start = pd.NaT
            real_end   = pd.NaT
            entries = it.get("entries") or []
            if isinstance(entries, list) and entries:
                for e in entries:
                    if isinstance(e, dict) and e.get("type") == "START" and pd.isna(real_start):
                        real_start = iso_to_dt_ar(e.get("time") or e.get("date"))
                        break
                for e in entries:
                    if isinstance(e, dict) and e.get("type") == "END":
                        real_end = iso_to_dt_ar(e.get("time") or e.get("date"))

            reg, ext = hours_regular_extra(it.get("categorizedHours"))
            obs = build_observaciones(it)

            rows.append({
                "ID": emp,
                "APELLIDO, NOMBRE": f"{u.get('Apellido','')}, {u.get('Nombre','')}".strip(", ").strip(),
                "FECHA": ref,
                "DIA": weekday_es(ref),

                "_scheduledStart": sched_start,
                "_scheduledEnd": sched_end,
                "_realStart": real_start,
                "_realEnd": real_end,

                "HORARIO_OBLIGATORIO": fmt_range(sched_start, sched_end),
                "FICHADAS": fmt_range(real_start, real_end),

                "OBSERVACIONES": obs,

                "HORAS_REGULARES": round(reg, 2),
                "HORAS_EXTRA": round(ext, 2),
            })

        if len(items) < LIMIT_DAYS:
            break
        page += 1

    return rows

# ================= EXEC =================
all_rows = []
with ThreadPoolExecutor(max_workers=MAX_WORKERS) as ex:
    futures = [
        ex.submit(fetch_batch, employee_ids[i:i+BATCH_SIZE])
        for i in range(0, len(employee_ids), BATCH_SIZE)
    ]
    for f in as_completed(futures):
        all_rows.extend(f.result())

df = pd.DataFrame(all_rows)
print(f"TOTAL FILAS: {len(df)}")

# ================= CALCULOS =================
df["TARDANZA_HORAS"] = (
    (df["_realStart"] - df["_scheduledStart"])
    .dt.total_seconds().div(3600)
    .clip(lower=0)
    .round(2)
)

df["ANTICIPADO_HORAS"] = (
    (df["_scheduledStart"] - df["_realStart"])
    .dt.total_seconds().div(3600)
    .clip(lower=0)
    .round(2)
)

df["RETIRO_ANTICIPADO_HORAS"] = (
    (df["_scheduledEnd"] - df["_realEnd"])
    .dt.total_seconds().div(3600)
    .clip(lower=0)
    .round(2)
)

# ================= FLAG DISPLAY (HH:MM vs decimal) =================
if MOSTRAR_METRICAS_EN_HHMM:
    df["TARDANZA"] = df["TARDANZA_HORAS"].apply(horas_a_hhmm)
    df["ANTICIPADO"] = df["ANTICIPADO_HORAS"].apply(horas_a_hhmm)
    df["RETIRO_ANTICIPADO"] = df["RETIRO_ANTICIPADO_HORAS"].apply(horas_a_hhmm)

# ================= LIMPIEZA TECNICA =================
df = df.drop(columns=["_scheduledStart", "_scheduledEnd", "_realStart", "_realEnd"])

# ================= ORDEN FINAL =================
if MOSTRAR_METRICAS_EN_HHMM:
    cols = [
        "ID",
        "APELLIDO, NOMBRE",
        "FECHA",
        "DIA",
        "HORARIO_OBLIGATORIO",
        "FICHADAS",
        "OBSERVACIONES",
        "HORAS_REGULARES",
        "HORAS_EXTRA",
        "TARDANZA",
        "ANTICIPADO",
        "RETIRO_ANTICIPADO",
    ]
else:
    cols = [
        "ID",
        "APELLIDO, NOMBRE",
        "FECHA",
        "DIA",
        "HORARIO_OBLIGATORIO",
        "FICHADAS",
        "OBSERVACIONES",
        "HORAS_REGULARES",
        "HORAS_EXTRA",
        "TARDANZA_HORAS",
        "ANTICIPADO_HORAS",
        "RETIRO_ANTICIPADO_HORAS",
    ]

df = df[cols]

# ================= EXCEL =================
out = f"reporte_tardanzas_{START_DATE}_a_{END_DATE}.xlsx"
df.to_excel(out, index=False)
print("OK → Excel generado:", out)
```
